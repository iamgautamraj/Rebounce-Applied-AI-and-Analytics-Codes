# Track B - Text Article Analyzer
This track aims to build a tool to analyze the passage/comprehensions to get insights like word counts, total sentences, reading time etc. It involves concepts of functions, strings, dictionaries, sorting with lambda, file I/O.
Pure python code has been used with no libraries imported.

## How to run it?


1.   Upload any .txt file in `Files` in `Google Colab`. Rename it to `text.txt`.
2.   Run all cells from top to bottom sequentially.
3.   To adjust the stopwords, edit the list in `cell 4`.
4.   Running the last cell generates a new file `output.txt` with the required output.



```python
def takeinput():
  with open("text.txt", "r", encoding="utf-8") as file:
    text_content = file.read()
  return text_content
```

```python
def countwords(string):
  count=len(string.split())
  return count
#print(count)
```

```python
def sentenceCount(string):
  sentences=sum(1 for char in string if char in ".!?")
  return sentences
```

```python
def topten(string):
  data={}
  stopwords=['the','and','or','a','an','if','else','to','for','of','in','on','but','from','it','with','is','was','were','this','that','as',"no","yes","are",'at','we','you','I','he','she','me','de','be','us','The','ad','so','They','they','all','A','a','do','does','has','have','what','one','here','there','out','can','could','some','most','much','how','why','then']
  for char in string.split():
    if char in stopwords:
      continue
    cnt=string.count(char)
    data[char]=cnt
  sorted_data=dict(sorted(data.items(), key=lambda item: item[1], reverse=True))
  top10=list(sorted_data.items())[:10]
  return top10
#print(data)
```

```python
def averageLen(string, totalwords):
  length=0
  for char in string.split():
    length+=len(char)
  averageLen=length/totalwords
  return averageLen
```

```python
def readingtime(totalwords):
  return totalwords/200
```

```python
def main():
  string=takeinput()
  totalwords=countwords(string)
  totalsentences=sentenceCount(string)
  #topten=topten(string)
  averageWordLength=averageLen(string, totalwords)
  toptenwords=topten(string)
  readingtime=totalwords/200

  print("=== Article Analysis ===")
  print("Word Count:",totalwords)
  print("Sentence count:",totalsentences)
  print("Average word length:",round(averageWordLength,2))
  print("Reading time:","~",round(readingtime),"minutes")

  print("Top 10 most common words:")
  id=1
  for x,y in toptenwords:
    print(f"{id}. {x}: {y}")
    id+=1

  with open("output.txt", "w", encoding="utf-8") as file:
    file.write("=== Article Analysis ===\n")
    file.write(f"Word Count:   {totalwords}\n")
    file.write(f"Sentence count:   {totalsentences}\n")
    file.write(f"Average word length: {round(averageWordLength, 2)}\n")
    file.write(f"Reading time:   ~{round(readingtime)} minutes\n")

    file.write("\nTop 10 most common words:\n")
    id = 1
    for x, y in toptenwords:
        file.write(f"{id}. {x}: {y}\n")
        id += 1
```

```python
main()
```
